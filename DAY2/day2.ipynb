```python
n = int(input("Enter N: "))
number = int(input("Enter a number: "))
print("Primes:", end=" ")
for i in range(2, n + 1):
    prime = True
    for j in range(2, i):
        if i % j == 0:
            prime = False
            break
    if prime:
        print(i, end=" ")
factorial = 1
for i in range(1, n + 1):
    factorial *= i
print("\nFactorial:", factorial)
a, b = 0, 1
print("Fibonacci:", end=" ")
for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
temp = abs(number)
digit_sum = 0
reverse = 0

while temp > 0:
    digit = temp % 10
    digit_sum += digit
    reverse = reverse * 10 + digit
    temp //= 10

print("\nDigit Sum:", digit_sum)
print("Reverse:", reverse)
```

```python
n = int(input("Enter pattern size N: "))
print("\nRight Triangle")
for i in range(1, n + 1):
    for j in range(i):
        print("*", end="")
    print()

print("\nPyramid")
for i in range(1, n + 1):
    for j in range(n - i):
        print(" ", end="")
    for j in range(2 * i - 1):
        print("*", end="")
    print()

print("\nNumber Triangle")
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end="")
    print()

print("\nMultiplication Grid")
for i in range(1, 11):
    for j in range(1, 11):
        print(f"{i} x {j} = {i*j:<3}", end="  ")
    print()
```

```python
text = input("Enter a string: ")
reverse_text = ""
for ch in text:
    reverse_text = ch + reverse_text

print("Loop Reverse:", reverse_text)
print("Slice Reverse:", text[::-1])

# Palindrome ignoring spaces and case
cleaned = ""
for ch in text:
    if ch != " ":
        cleaned += ch.lower()

print("Palindrome:", "Yes" if cleaned == cleaned[::-1] else "No")

vowels = 0
consonants = 0
digits = 0

for ch in text.lower():
    if ch in "aeiou":
        vowels += 1
    elif ch.isalpha():
        consonants += 1
    elif ch.isdigit():
        digits += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
```

```python
name = input("Enter customer name: ")
units = int(input("Enter units consumed: "))

if units < 0:
    print("Invalid units")
else:
    # Example slab rates
    if units <= 100:
        amount = units * 2.50
    elif units <= 200:
        amount = 100 * 2.50 + (units - 100) * 4
    else:
        amount = 100 * 2.50 + 100 * 4 + (units - 200) * 6

    print("\n----- Electricity Bill -----")
    print(f"Customer : {name}")
    print(f"Units    : {units}")
    print(f"Amount   : Rs.{amount:.2f}")
```

```python
amount = int(input("Enter withdrawal amount: "))

if amount <= 0:
    print("Invalid withdrawal amount")
elif amount > 20000:
    print("Withdrawal limit exceeded")
elif amount % 10 != 0:
    print("Amount must be a multiple of 10")
else:
    remaining = amount
    total_notes = 0

    for denomination in [500, 200, 100, 50, 10]:
        notes = remaining // denomination
        remaining %= denomination
        total_notes += notes
        print(f"Rs.{denomination} notes: {notes}")

    print("Total Notes:", total_notes)
    print("Amount:", amount)
```

```python
entry = input("Enter entry time (HH:MM): ")
exit_time = input("Enter exit time (HH:MM): ")

try:
    eh, em = map(int, entry.split(":"))
    xh, xm = map(int, exit_time.split(":"))

    if not (0 <= eh < 24 and 0 <= em < 60 and
            0 <= xh < 24 and 0 <= xm < 60):
        print("Invalid time")
    else:
        start = eh * 60 + em
        end = xh * 60 + xm

        if end < start:
            end += 24 * 60

        duration = end - start
        hours = duration // 60
        minutes = duration % 60
        billable_hours = (duration + 59) // 60

        if duration == 0:
            fee = 0
        elif billable_hours == 1:
            fee = 30
        else:
            fee = 30 + (billable_hours - 1) * 20

        print(f"Parking Duration: {hours} hours {minutes} minutes")
        print("Billable Hours:", billable_hours)
        print("Parking Fee: Rs.", fee)

except ValueError:
    print("Invalid time format. Use HH:MM.")

```

```python

print("\n========== Q8: Reverse Integer ==========")

x = int(input("Enter an integer: "))

sign = -1 if x < 0 else 1
temp = abs(x)
reverse = 0

while temp > 0:
    digit = temp % 10
    reverse = reverse * 10 + digit
    temp //= 10

print("Reversed Integer:", sign * reverse)
```
