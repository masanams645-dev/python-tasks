```python

stock = {'LED': 20, 'RESISTOR': 50, 'SENSOR': 4}
requested = ['LED', 'SENSOR', 'SWITCH']

for item in requested:
    if item in stock:
        print(f"{item}: {stock[item]} available")
    else:
        print(f"{item}: unavailable")

```

```python

store_a = {'LED': 20, 'Cable': 10, 'Board': 3}
store_b = {'Cable': 5, 'Sensor': 7, 'LED': 8}

combined = store_a.copy()

for item, quantity in store_b.items():
    combined[item] = combined.get(item, 0) + quantity

print("Combined inventory:", combined)
```

```python

sales = {
    'Sensor': [4, 6, 5],
    'Cable': [10, 2, 3],
    'Board': [8, 7],
    'LED': [5, 5, 5]
}

totals = {product: sum(values)
          for product, values in sales.items()}

ranked = sorted(totals.items(),
                key=lambda x: (-x[1], x[0]))

for product, total in ranked:
    print(f"{product}: {total}")
```

```python

permissions = {
    'admin': {'read', 'write', 'delete'},
    'operator': {'read', 'write'},
    'auditor': {'read', 'export'}
}

sets = list(permissions.values())

common = set.intersection(*sets)
all_permissions = set.union(*sets)

print("Common permissions:", sorted(common))
print("All permissions:", sorted(all_permissions))
```

```python

def subarraySum(nums, k):
    prefix_count = {0: 1}
    prefix_sum = 0
    count = 0

    for num in nums:
        prefix_sum += num
        count += prefix_count.get(prefix_sum - k, 0)
        prefix_count[prefix_sum] = (
            prefix_count.get(prefix_sum, 0) + 1
        )

    return count


nums = [1, 1, 1]
k = 2

print("Count:", subarraySum(nums, k))
```

```python

from collections import Counter

def minWindow(s, t):
    if not t or len(t) > len(s):
        return ""

    need = Counter(t)
    window = {}
    left = 0
    required = len(need)
    formed = 0
    min_len = float('inf')
    start = 0

    for right, char in enumerate(s):
        window[char] = window.get(char, 0) + 1

        if char in need and window[char] == need[char]:
            formed += 1

        while formed == required:
            if right - left + 1 < min_len:
                min_len = right - left + 1
                start = left

            left_char = s[left]
            window[left_char] -= 1

            if (left_char in need and
                    window[left_char] < need[left_char]):
                formed -= 1

            left += 1

    if min_len == float('inf'):
        return ""

    return s[start:start + min_len]


s = 'ADOBECODEBANC'
t = 'ABC'

print("Minimum window:", repr(minWindow(s, t)))
```
