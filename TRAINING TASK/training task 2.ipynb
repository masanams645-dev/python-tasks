```python
def product_except_self(nums):
    n = len(nums)
    answer = [1] * n

    prefix = 1
    for i in range(n):
        answer[i] = prefix
        prefix *= nums[i]

    suffix = 1
    for i in range(n - 1, -1, -1):
        answer[i] *= suffix
        suffix *= nums[i]

    return answer


nums = [1, 2, 3, 4]
print("Product array:", product_except_self(nums))
```

```python
def calculate_bill(prices, tax_rate=0.10, discount=0.20):
    subtotal = sum(prices)
    discount_amount = subtotal * discount
    taxable_amount = subtotal - discount_amount
    tax = taxable_amount * tax_rate
    total = taxable_amount + tax

    print(f"Subtotal: {subtotal:.2f}")
    print(f"Discount: {discount_amount:.2f}")
    print(f"Tax: {tax:.2f}")
    print(f"Final total: {total:.2f}")


prices = [100, 50]
calculate_bill(prices, tax_rate=0.10, discount=0.20)
```

```python
def average_readings(*args):
    if len(args) == 0:
        return None

    return sum(args) / len(args)


result = average_readings(10, 20, 30, 40)

if result is not None:
    print("Average:", result)
else:
    print("No readings provided")

result = average_readings()

if result is not None:
    print("Average:", result)
else:
    print("No readings provided")
```

```python
def create_device_config(device_id, **kwargs):
    allowed = ["baud_rate", "parity", "timeout_ms", "debug"]

    for key in kwargs:
        if key not in allowed:
            raise ValueError(f"Unsupported key: {key}")

    config = {"device_id": device_id}
    config.update(kwargs)

    return config


result = create_device_config(
    "D01",
    baud_rate=115200,
    parity="N",
    timeout_ms=500
)

print(result)
```

```python
def group_anagrams(strs):
    groups = {}

    for word in strs:
        key = "".join(sorted(word))

        if key not in groups:
            groups[key] = []

        groups[key].append(word)

    return list(groups.values())


strs = ["eat", "tea", "tan", "ate", "nat", "bat"]

print("Groups:")
for group in group_anagrams(strs):
    print(group)
```

```python
def top_k_frequent(nums, k):
    frequency = {}

    for num in nums:
        frequency[num] = frequency.get(num, 0) + 1

    sorted_nums = sorted(
        frequency,
        key=lambda num: frequency[num],
        reverse=True
    )

    return sorted_nums[:k]


nums = [1, 1, 1, 2, 2, 3]
k = 2

print("Top", k, "frequent values:", top_k_frequent(nums, k))
```

```python
def longest_consecutive(nums):
    num_set = set(nums)
    longest = 0
    best_sequence = []

    for num in num_set:
        if num - 1 not in num_set:
            current = num
            length = 1

            while current + 1 in num_set:
                current += 1
                length += 1

            if length > longest:
                longest = length
                best_sequence = list(range(num, current + 1))

    return longest, best_sequence


nums = [100, 4, 200, 1, 3, 2]
length, sequence = longest_consecutive(nums)

print("Longest consecutive sequence length:", length)
print("Sequence:", ", ".join(map(str, sequence)))
```

```python
def find_max_length(nums):
    balance = 0
    first_seen = {0: -1}
    max_length = 0

    for i, num in enumerate(nums):
        if num == 0:
            balance -= 1
        else:
            balance += 1

        if balance in first_seen:
            max_length = max(max_length, i - first_seen[balance])
        else:
            first_seen[balance] = i

    return max_length


nums = [0, 1, 0]
print("Maximum balanced subarray length:", find_max_length(nums))
```

```python
def length_of_last_word(s):
    words = s.split()
    return len(words[-1]) if words else 0


s = "   Embedded Python   "
print("Length of last word:", length_of_last_word(s))
```

```python
def str_str(haystack, needle):
    if needle == "":
        return 0

    for i in range(len(haystack) - len(needle) + 1):
        if haystack[i:i + len(needle)] == needle:
            return i

    return -1


haystack = "datascience"
needle = "science"

print("First occurrence index:", str_str(haystack, needle))
```
