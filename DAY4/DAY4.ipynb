```python

# Q1. Number Classification Report

n = int(input())
numbers = list(map(int, input().split()))

positive = negative = zero = 0
even = odd = 0
both = only3 = only5 = neither = 0

for num in numbers:
    if num > 0:
        positive += 1
    elif num < 0:
        negative += 1
    else:
        zero += 1

    if num != 0:
        if num % 2 == 0:
            even += 1
        else:
            odd += 1

    if num % 3 == 0 and num % 5 == 0:
        both += 1
    elif num % 3 == 0:
        only3 += 1
    elif num % 5 == 0:
        only5 += 1
    else:
        neither += 1

print("Positive:", positive)
print("Negative:", negative)
print("Zero:", zero)
print("Even:", even)
print("Odd:", odd)
print("Divisible by both 3 and 5:", both)
print("Divisible only by 3:", only3)
print("Divisible only by 5:", only5)
print("Divisible by neither:", neither)
```

```python
# Q2. Valid Input Until Accepted

invalid_attempts = 0

while True:
    num = int(input())

    if num < 1 or num > 100:
        print("Invalid input")
        invalid_attempts += 1
        continue

    break

print("Accepted value:", num)
print("Invalid attempts:", invalid_attempts)
```

```python

# Q4. Multiplication Table with Selective Skipping

n = int(input())
skipped = 0

for i in range(1, 21):
    result = n * i

    if result > 100:
        break

    if result % 3 == 0:
        skipped += 1
        continue

    print(f"{n} x {i} = {result}")

print("Skipped results:", skipped)

```

```python
# Q5. Password Validation System

password = input()

uppercase = False
lowercase = False
digit = False
special = False

for ch in password:
    if ch.isupper():
        uppercase = True
    elif ch.islower():
        lowercase = True
    elif ch.isdigit():
        digit = True
    else:
        special = True

length_valid = len(password) >= 8

if uppercase:
    print("Uppercase: Present")
else:
    print("Uppercase: Missing")

if lowercase:
    print("Lowercase: Present")
else:
    print("Lowercase: Missing")

if digit:
    print("Digit: Present")
else:
    print("Digit: Missing")

if special:
    print("Special character: Present")
else:
    print("Special character: Missing")

if length_valid:
    print("Minimum length: Satisfied")
else:
    print("Minimum length: Not satisfied")

if uppercase and lowercase and digit and special and length_valid:
    print("Valid password")
else:
    print("Invalid password")
```

```python
# Q6. Hollow Square Pattern

n = int(input())

for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# Task 7: First Perfect Square in a Range

L, R = map(int, input("Enter L and R: ").split())

found = False

for i in range(L, R + 1):
    for j in range(1, i + 1):
        if j * j == i:
            print("First perfect square:", i)
            found = True
            break
        elif j * j > i:
            break

    if found:
        break

if not found:
    print("No perfect square found")
```

```python
# Q8. Number Guessing Game with Limited Attempts

secret = 8
attempts = 0

while attempts < 5:
    guess = int(input())
    attempts += 1

    if guess == secret:
        print("Correct")
        break
    elif guess < secret:
        print("Too Low")
    else:
        print("Too High")

if attempts == 5 and guess != secret:
    print("Game Over")

```

```python
# Task 9: Longest Consecutive Equal Values

n = int(input("Enter the number of elements: "))
numbers = list(map(int, input("Enter the elements: ").split()))

current_value = numbers[0]
current_length = 1

best_value = numbers[0]
best_length = 1

for i in range(1, n):
    if numbers[i] == current_value:
        current_length += 1
    else:
        current_value = numbers[i]
        current_length = 1

    if current_length > best_length:
        best_length = current_length
        best_value = current_value

print("Value =", best_value)
print("Length =", best_length)
```

```python
# Task 10: Transaction Validation & Summary

normal = 0
large = 0
suspicious = 0
total = 0

while True:
    amount = int(input("Enter transaction amount: "))

    if amount == -1:
        break

    if amount == 0:
        pass
    elif amount < 0:
        continue
    elif amount <= 1000:
        normal += 1
        total += amount
    elif amount <= 5000:
        large += 1
        total += amount
    else:
        suspicious += 1
        total += amount

        if suspicious == 3:
            break

print("Normal transactions:", normal)
print("Large transactions:", large)
print("Suspicious transactions:", suspicious)
print("Total valid amount:", total)
```
